```python
import pandas as pd
import duckdb
```

```python
FILENAME = "entities_to_concepts.tsv"

from urllib.request import urlretrieve # would prefer to just !wget URL but didn't set up this laptop with jupyter linked to powershell or WSL

url = "https://github.com/elixir-europe/micoknowgraph/raw/a411abb32e97060a6333c2ec184bc1542e4eb026/data/entity_mapping/entities_to_concepts.tsv"
urlretrieve(url, "entities_to_concepts.tsv")


df = pd.read_csv( FILENAME, sep="\t", header=0) 
```

```python
display(df)
```

```python
# Pull rows that contain 'gene'

# 1. Create a boolean mask across all columns
mask = df.apply(lambda col: col.astype(str).str.contains(r'gene', case=False, regex=True, na=False))

# 2. Filter the DataFrame to get the matching rows
matching_rows = df[mask.any(axis=1)]

# 3. Display those rows
display(matching_rows)
```

### Pull rows that concept_id is blank

```python
empty_rows = df[df['concept_id'].isna()]
display(empty_rows)
```

### Pull rows that concept_id column is blank and kind column does not have 'measurement', 'statistic', 'meta', 'method-or-procedure', 'data-artifact' and map column is equal to yes

```python
empty_rows = df[df['concept_id'].isna()& ~df['kind'].isin(['measurement', 'statistic','meta','method-or-procedure','data-artifact']) & (df['map'] == 'yes')]
display(empty_rows)
```
